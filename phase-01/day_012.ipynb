```python
plt.figure(figsize=(15, 6))

# Plot 1: Age distribution of users
plt.subplot(1, 2, 1)
sns.histplot(data=data, x='age', bins=30, kde=True, color='teal')
plt.title('Age Distribution of Users')
plt.xlabel('Age')
plt.ylabel('Count')

# Plot 2: Average likes received by gender
plt.subplot(1, 2, 2)
sns.barplot(x='gender', y='likes_received', data=data.dropna(subset=['gender']), palette='viridis')
plt.title('Average Likes Received by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Likes Received')

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

clf_data = data.dropna(subset=['gender']).copy()

features = ['age', 'friend_count', 'friendships_initiated', 'likes', 'likes_received']
X = clf_data[features]
y = clf_data['gender']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

classifier = RandomForestClassifier(n_estimators=50, max_depth=10, random_state=42, n_jobs=-1)
classifier.fit(X_train, y_train)

y_pred = classifier.predict(X_test)

cm = confusion_matrix(y_test, y_pred, labels=classifier.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=classifier.classes_)

fig, ax = plt.subplots(figsize=(8, 6))
disp.plot(cmap='Blues', ax=ax)
plt.title('Confusion Matrix: Predicting Gender from Social Metrics')
plt.show()
```

```python
plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
sns.histplot(data=data, x='tenure', bins=30, color='salmon', kde=True)
plt.title('Distribution of User Tenure (Days)')
plt.xlabel('Tenure (Days)')
plt.ylabel('Count')

sample_data = data.sample(2000, random_state=42)
plt.subplot(1, 3, 2)
sns.scatterplot(data=sample_data, x='age', y='friend_count', alpha=0.5, hue='gender', palette='Set1')
plt.title('Age vs Friend Count (Sample)')
plt.xlabel('Age')
plt.ylabel('Friend Count')

plt.subplot(1, 3, 3)
likes_comparison = pd.DataFrame({
    'Like Source': ['Mobile Likes', 'Web Likes'],
    'Total Likes': [data['mobile_likes'].sum(), data['www_likes'].sum()]
})
sns.barplot(x='Like Source', y='Total Likes', data=likes_comparison, palette='coolwarm', hue='Like Source', legend=False)
plt.title('Total Likes: Mobile vs Web')
plt.xlabel('Platform')
plt.ylabel('Total Likes (in Millions)')

plt.tight_layout()
plt.show()
```
